## Rule Chọn Trường Cho Data Seed

### 1. Phân loại category

Chọn **index_group_name + product_group_name → category**

Ví dụ:

```text
index_group_name = Ladieswear
product_group_name = Garment Upper body
```

Được ánh xạ thành:

```text
Áo Nữ
```

- **Ladieswear** xác định nhóm thời trang nữ.
- **Garment Upper body** xác định sản phẩm thuộc phần thân trên.

---

### 2. Các specification được chọn

| Specification | Trường dữ liệu | Giá trị |
|---|---|---|
| Loại sản phẩm | **product_type_name** | **Vest top** |
| Họa tiết | **graphical_appearance_name** | **Solid** |
| Dòng hàng | **garment_group_name** | **Jersey Basic** |
| Bộ sưu tập | **section_name** | **Womens Everyday Basics** |

- **product_type_name** mô tả loại sản phẩm cụ thể.
- **graphical_appearance_name** mô tả họa tiết hoặc bề mặt.
- **garment_group_name** mô tả nhóm dòng hàng/chất liệu.
- **section_name** mô tả phân khúc hoặc bộ sưu tập.

Chọn các trường dữ liệu này vì chúng là text và phù hợp với specification của sản phẩm.

---

## 2. Missing values, duplicates và độ dài trường

Các cột được kiểm tra là những cột được seed sử dụng

````python
SEEDED_COLUMNS = [
    "article_id",
    "product_code",
    "prod_name",
    "product_type_name",
    "product_group_name",
    "graphical_appearance_name",
    "colour_group_name",
    "index_group_name",
    "section_name",
    "garment_group_name",
    "detail_desc",
]

available_seeded_columns = [
    column
    for column in SEEDED_COLUMNS
    if column in df.columns
]

missing_columns = [
    column
    for column in SEEDED_COLUMNS
    if column not in df.columns
]

if missing_columns:
    print("Missing expected CSV columns:", missing_columns)

seeded_df = df[available_seeded_columns]

null_count = seeded_df.isna().sum()

empty_string_count = (
    seeded_df
    .astype("string")
    .apply(lambda column: column.str.strip().eq("").sum())
)

missing = (
    pd.DataFrame({
        "null_count": null_count,
        "empty_string_count": empty_string_count,
    })
    .assign(
        missing_count=lambda x: (
            x["null_count"] + x["empty_string_count"]
        ),
        missing_percent=lambda x: (
            x["missing_count"] / len(seeded_df) * 100
        ),
    )
    .sort_values("missing_count", ascending=False)
)

display(missing)
````

## 3. Taxonomy và phân bố category nguồn

````python
taxonomy_columns = [
    "index_group_name",
    "product_group_name",
    "product_type_name",
    "section_name",
    "garment_group_name",
]

# Một dòng đại diện cho một product_code.
# Vì các article cùng product_code có chung taxonomy.
product_level_df = (
    df
    .sort_values(["product_code", "article_id"])
    .drop_duplicates("product_code")
)

for column in taxonomy_columns:
    article_counts = (
        df[column]
        .value_counts(dropna=False)
        .rename("article_count")
    )

    product_counts = (
        product_level_df[column]
        .value_counts(dropna=False)
        .rename("product_count")
    )

    counts = (
        pd.concat([article_counts, product_counts], axis=1)
        .fillna(0)
        .astype(int)
        .sort_values("product_count", ascending=False)
    )

    print(f"\n### {column}")
    display(counts.head(30))
````

````python
article_taxonomy_matrix = pd.crosstab(
    df["index_group_name"],
    df["product_group_name"],
)
display(article_taxonomy_matrix)
````

````python
product_taxonomy_matrix = pd.crosstab(
    product_level_df["index_group_name"],
    product_level_df["product_group_name"],
)

display(product_taxonomy_matrix)
````


## Kết luận

Tạo 12 category dựa trên các tổ hợp:

`index_group_name` :
Accessories
Garment Lower body
Garment Upper body

và

`product_group_name` 

```text
Baby/Children
Ladieswear
Menswear
Sport

````python
TARGET_INDEX_GROUPS = [
    "Baby/Children",
    "Ladieswear",
    "Menswear",
    "Sport",
]

TARGET_PRODUCT_GROUPS = [
    "Accessories",
    "Garment Lower body",
    "Garment Upper body",
]

filtered_df = df[
    df["index_group_name"].isin(TARGET_INDEX_GROUPS)
    & df["product_group_name"].isin(TARGET_PRODUCT_GROUPS)
]

product_colourways = (
    filtered_df
    .groupby(
        [
            "index_group_name",
            "product_group_name",
            "product_code",
        ],
        dropna=False,
    )["colour_group_name"]
    .nunique()
    .reset_index(name="colour_count")
)

products_over_n_colours = product_colourways[
    product_colourways["colour_count"] > 1
]

summary = (
    products_over_n_colours
    .groupby(
        [
            "index_group_name",
            "product_group_name",
        ]
    )
    .size()
    .unstack(fill_value=0)
    .reindex(
        index=TARGET_INDEX_GROUPS,
        columns=TARGET_PRODUCT_GROUPS,
        fill_value=0,
    )
)

summary
````
